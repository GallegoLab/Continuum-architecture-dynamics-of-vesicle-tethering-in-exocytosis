```python
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib as mpl

mpl.rcParams['font.size']  = 18
mpl.rcParams['axes.linewidth'] = 1.5
mpl.rcParams['axes.labelsize'] = 18
mpl.rcParams['xtick.labelsize'] = 16
mpl.rcParams['ytick.labelsize'] = 17
pd.options.display.float_format = '{:,.4f}'.format
```

```python
# Representative mosaic of simulated sites
home = "../" # Update for the proper location of SMLM folder
alldata= pd.read_csv(home + "Data/Synthetic_SMLM_clusters.csv")
ids = np.random.randint(len(alldata))
fig,axes = plt.subplots(3,3,figsize=(3*3,3*3))
axes  = np.ravel(axes)
for idf, ids in enumerate(alldata.sample(9).index):
    if idf < 9:
        ax = axes[idf]
        for i,j in enumerate(["Locs","Exos"]):
            x = alldata.loc[ids, "X_"+j]
            x = x.split('[')[-1].split(']')[0].split("\n")
            x = [float(xxx) for xx in x for xxx in xx.split(" ") if len(xxx) > 0 ]
            y = alldata.loc[ids, "Y_"+j]
            y = y.split('[')[-1].split(']')[0].split("\n")
            y = [float(yyy) for yy in y for yyy in yy.split(" ") if len(yyy) > 0 ]

            ro = 35-15
            r1 = 35+15
            th = np.linspace(0,1,50)*2*np.pi
            ax.scatter(x,y,color=['k','b'][i])        
        
        ax.plot(r1*np.cos(th),r1*np.sin(th),lw=3,color='r')
        ax.plot(ro*np.cos(th),ro*np.sin(th),lw=3,color='r')
        ax.set_xlim(-150,150)
        ax.set_ylim(-150,150)
        ax.set_xticks([])
        ax.set_yticks([])
```
